```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("Nat_Gas.csv")
data.head()
```

```python
data["Date"] = pd.to_datetime(data["Date"])
```

```python
data.head()
```

```python
plt.plot(data["Date"], data["Price"])
```

```python
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA
```

```python
def adftest(x):
    result = adfuller(x)
    print("ADFStatistics : " , result[0])
    print("p_value: ", result[1])

    if(result[1] < 0.5):
        print("Stationary")
    else:
        print("Not Stationary")
        
adftest(data["Price"])
```

```python
series_diff = data["Price"].diff().dropna()

print("\nADF Test After First Differencing:")
adftest(series_diff)
```

```python
model = ARIMA(data["Price"], order = (1,1,12))
model_fit = model.fit()

print(model_fit.summary())

forecast = model_fit.forecast(steps=12)

    
```

```python

plt.plot(data["Price"], label="Historical")
plt.plot(forecast, label="1-Year Forecast")
plt.legend()
plt.title("Natural Gas Price Forecast (ARIMA)")
plt.show()

print("\nForecasted Values:")
print(forecast)
```
